# Notebook 7 — Demand Forecasting

## Objective

This notebook develops a time-series forecasting framework for future retail demand.

The core business question is:

> **How much demand should we expect in the future, and where will that demand come from?**

Unlike the earlier customer-level notebooks, this notebook focuses on demand over time rather than individual customer behaviour.

The objective is to forecast future commercial activity using only information that would have been available at the historical forecasting date.

---

## Why Demand Forecasting Matters

Demand forecasting supports decisions such as:

- inventory planning;
- purchasing and replenishment;
- working-capital allocation;
- staffing and fulfilment planning;
- revenue planning;
- product prioritisation;
- and identification of unusually strong or weak trading periods.

A useful forecasting system should not only estimate future demand, but also make clear:

- the forecasting horizon;
- the level at which demand is being predicted;
- the uncertainty around the forecast;
- and whether the forecast is sufficiently stable for operational use.

---

## Forecasting Scope

This notebook will begin with demand at the broader business level before considering more granular forecasting.

The analysis will distinguish between several possible forecasting targets:

### Revenue Demand

Forecast future qualifying merchandise revenue.

### Order Demand

Forecast the number of qualifying customer orders.

### Unit Demand

Forecast the quantity of merchandise purchased.

### Product-Level Demand

Forecast future demand for individual products or product groups where sufficient history exists.

Not every product is expected to support a reliable standalone time-series model.

Sparse and intermittent products will therefore be treated differently from high-volume products.

---

## Forecasting Granularity

The first forecasting level will be:

> **weekly demand**

Weekly aggregation provides a balance between:

- retaining meaningful commercial variation;
- reducing daily noise;
- producing enough historical observations for temporal modelling;
- and remaining operationally useful for planning.

Monthly summaries may also be used for strategic interpretation.

Daily forecasting will not be the primary objective because the dataset contains substantial day-to-day volatility and irregular trading patterns.

---

## Primary Forecast Target

The primary initial forecasting target will be:

> **Weekly qualifying positive merchandise revenue**

using the same qualifying merchandise definition applied throughout the project.

This represents revenue from identified and unidentified positive merchandise purchase activity while excluding:

- cancellations;
- returns;
- non-merchandise accounting adjustments;
- zero-price transactions;
- and observations beyond the complete-data boundary.

Additional demand measures such as orders and units will be analysed as supporting signals.

---

## Important Revenue Interpretation

The forecasting target represents revenue from qualifying positive merchandise transactions.

It should not be interpreted as an accounting forecast of final recognised revenue after all future returns, cancellations, refunds or accounting adjustments.

The forecasting system is therefore intended primarily for:

> **commercial demand planning**

rather than statutory accounting forecasting.

---

## Observation Boundary

The dataset contains only a partial December 2011 period.

Therefore:

> **30 November 2011**

will remain the final complete observation date for demand modelling.

Partial December 2011 data will not be used as a normal complete forecasting period.

---

## Forecast Horizons

Forecast quality will be evaluated across multiple operational horizons.

Initial horizons will include:

- **1 week ahead**
- **2 weeks ahead**
- **4 weeks ahead**

The 4-week horizon is particularly important because it approximates one month of forward planning.

Forecast performance generally deteriorates as the horizon increases, so each horizon will be evaluated separately.

---

## Forecasting Philosophy

All forecasting evaluation will be strictly temporal.

Random train/test splitting is inappropriate for this problem.

Each forecast must be generated using only observations that occurred before the forecast origin.

The notebook will therefore use:

> **rolling-origin / expanding-window validation**

where models are repeatedly trained on historical data and evaluated on later unseen periods.

This more closely represents how the forecasting system would operate in production.

---

## Benchmark First

Sophisticated forecasting models are useful only if they outperform simple business rules.

The notebook will therefore begin with strong forecasting baselines, including:

### Naive Forecast

> Next period = most recent observed period

### Seasonal Naive Forecast

> Next period = demand from the corresponding historical seasonal period

where sufficient seasonal history exists.

### Moving Average Forecast

> Forecast based on recent average demand

### Historical Trend Baseline

A simple trend-based forecast using historical demand movement.

More complex models will only be retained if they provide meaningful out-of-time improvement over these baselines.

---

## Candidate Forecasting Models

Depending on the characteristics of the demand series, later candidate models may include:

- Exponential Smoothing;
- Holt trend models;
- Holt-Winters seasonal models;
- autoregressive models;
- regression with calendar and lag features;
- Gradient Boosting with time-series features;
- and specialised intermittent-demand methods for sparse products.

Model complexity will be justified by out-of-time performance rather than assumed in advance.

---

## Forecast Evaluation Metrics

No single forecasting metric is sufficient.

The notebook will evaluate models using several complementary measures.

### MAE — Mean Absolute Error

Measures typical forecast error in original demand units.

### RMSE — Root Mean Squared Error

Places more weight on large forecast misses.

### WAPE — Weighted Absolute Percentage Error

Measures total absolute forecast error relative to total actual demand.

This is particularly useful for business interpretation.

### Bias

Measures whether forecasts systematically overpredict or underpredict demand.

### Forecast / Actual Ratio

Compares aggregate predicted demand with aggregate realised demand.

### Directional and Peak Performance

Where useful, the analysis will also examine whether models correctly identify:

- demand acceleration;
- demand declines;
- and high-demand periods.

---

## Forecast Uncertainty

Point forecasts alone are insufficient for planning.

Where supported by the selected model, the final forecasting framework should also provide forecast intervals or planning ranges.

The goal is not to claim certainty.

The goal is to provide a defensible estimate of:

> **expected demand and the range of plausible outcomes.**

---

## Product-Level Forecasting

Product-level forecasting will not automatically be attempted for every SKU.

The catalogue contains a substantial long tail, and many products have insufficient historical volume for reliable independent forecasting.

Products will therefore be segmented into groups such as:

- high-volume / stable;
- medium-volume;
- low-volume;
- and intermittent demand.

Higher-volume products may support individual models.

Sparse products may require:

- pooled forecasts;
- category-level forecasts;
- historical-rate baselines;
- or intermittent-demand methods.

This avoids creating false precision for products with insufficient signal.

---

## Temporal Leakage Rules

No feature used for forecasting may contain information from the future.

This applies to:

- lagged revenue;
- rolling averages;
- order counts;
- unit counts;
- customer counts;
- product counts;
- calendar statistics;
- and any derived trend or seasonal feature.

Rolling features must always be shifted so that the current target period is not included in its own predictors.

---

## Test Governance

A final out-of-time forecasting period will be reserved before model selection.

The final Test period will not be used to:

- select forecasting algorithms;
- tune hyperparameters;
- select lag structures;
- choose rolling windows;
- or calibrate the final forecast.

Only after the forecasting architecture has been frozen will the final Test period be opened.

---

## Expected Notebook Outputs

By the end of this notebook, the project should produce:

1. a validated weekly demand series;
2. strong naive and seasonal benchmarks;
3. one or more candidate forecasting models;
4. rolling-origin validation results;
5. forecast-bias diagnostics;
6. forecast-interval or planning-range logic where appropriate;
7. a final out-of-time forecast evaluation;
8. demand forecasts by operational horizon;
9. guidance on which products support standalone forecasting;
10. and a final **GO / CONDITIONAL GO / NO-GO** deployment decision.

---

## Business Interpretation

The final forecasting system should answer:

> **What level of demand should we plan for over the next several weeks?**

It should not imply certainty.

Forecasts are conditional estimates based on historical patterns, and unexpected changes in:

- customer behaviour;
- product availability;
- pricing;
- promotions;
- economic conditions;
- or operating environment

may cause realised demand to differ materially from historical expectations.

---

## Notebook Governance Principle

Throughout this notebook:

> **historical prediction must resemble real future prediction.**

Forecasting performance will therefore be judged primarily by how well the model performs on later unseen periods rather than how well it explains historical data.


## Step 63 — Construct and Audit the Weekly Demand Series

This step constructs the weekly demand series that will be used throughout Notebook 7.

The primary forecasting target is:

> **Weekly qualifying positive merchandise revenue**

Supporting series include:

- weekly qualifying orders;
- weekly merchandise units;
- weekly active customers;
- weekly active products.

The purpose of this step is not to fit a forecasting model.

Instead, it establishes the time-series foundation and audits:

- date coverage;
- weekly continuity;
- missing weeks;
- partial weeks;
- demand distribution;
- seasonality;
- growth and volatility;
- and the relationship between revenue, orders and units.

The final complete-data boundary remains:

> **30 November 2011**


```python
# ================================================================
# 63.2 — IMPORTS
# ================================================================

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# ================================================================
# 63.3 — LOAD CANONICAL CLEANED DATA
# ================================================================

df = pd.read_parquet(
    "cleaned_online_retail.parquet"
)


print(
    f"Rows loaded: {len(df):,}"
)

print(
    f"Columns loaded: {df.shape[1]:,}"
)
```

```python
# ================================================================
# 63.4 — REQUIRED COLUMN VALIDATION
# ================================================================

required_columns = [
    "date",
    "invoice",
    "customer",
    "product",
    "quantity",
    "unit_price",
    "customer_is_identified",
    "is_merchandise",
    "is_cancellation",
    "net_merchandise_revenue",
]


missing_columns = [
    column
    for column in required_columns
    if column not in df.columns
]


assert not missing_columns, (
    f"Missing required columns: {missing_columns}"
)


print(
    "✓ Required demand-forecasting columns confirmed."
)
```

```python
# ================================================================
# 63.5 — COMPLETE-DATA BOUNDARY
# ================================================================

DEMAND_END_DATE = pd.Timestamp(
    "2011-11-30 23:59:59"
)


print(
    f"Demand observation boundary: "
    f"{DEMAND_END_DATE}"
)
```

```python
# ================================================================
# 63.6 — QUALIFYING POSITIVE MERCHANDISE
# ================================================================

demand_lines = (
    df.loc[
        df["is_merchandise"]
        &
        (df["quantity"] > 0)
        &
        (df["unit_price"] > 0)
        &
        (~df["is_cancellation"])
        &
        (df["date"] <= DEMAND_END_DATE)
    ]
    .copy()
)


print(
    f"Qualifying demand rows: "
    f"{len(demand_lines):,}"
)

print(
    f"Date range: "
    f"{demand_lines['date'].min()} "
    f"to "
    f"{demand_lines['date'].max()}"
)
```

```python
# ================================================================
# 63.7 — REVENUE RECONSTRUCTION AUDIT
# ================================================================

demand_lines[
    "reconstructed_revenue"
] = (
    demand_lines[
        "quantity"
    ]
    *
    demand_lines[
        "unit_price"
    ]
)


revenue_reconstruction_diff = (
    demand_lines[
        "net_merchandise_revenue"
    ]
    -
    demand_lines[
        "reconstructed_revenue"
    ]
)


print(
    "Maximum absolute revenue reconstruction difference:",
    f"{revenue_reconstruction_diff.abs().max():,.8f}"
)

print(
    "Total absolute reconstruction difference:",
    f"{revenue_reconstruction_diff.abs().sum():,.8f}"
)
```

```python
# ================================================================
# 63.8 — WEEKLY CALENDAR CONVENTION
# ================================================================

demand_lines[
    "week_period"
] = (
    demand_lines[
        "date"
    ]
    .dt.to_period(
        "W-SUN"
    )
)


demand_lines[
    "week_start"
] = (
    demand_lines[
        "week_period"
    ]
    .dt.start_time
)


demand_lines[
    "week_end"
] = (
    demand_lines[
        "week_period"
    ]
    .dt.end_time
)


print(
    demand_lines[
        [
            "date",
            "week_start",
            "week_end",
        ]
    ]
    .head()
)
```

```python
# ================================================================
# 63.9 — WEEKLY DEMAND AGGREGATION
# ================================================================

weekly_demand = (
    demand_lines
    .groupby(
        "week_period",
        as_index=False
    )
    .agg(
        revenue=(
            "net_merchandise_revenue",
            "sum"
        ),

        orders=(
            "invoice",
            "nunique"
        ),

        units=(
            "quantity",
            "sum"
        ),

        active_customers=(
            "customer",
            lambda values:
                values.dropna().nunique()
        ),

        active_products=(
            "product",
            "nunique"
        ),

        merchandise_lines=(
            "product",
            "size"
        ),
    )
)


weekly_demand[
    "week_start"
] = (
    weekly_demand[
        "week_period"
    ]
    .dt.start_time
)


weekly_demand[
    "week_end"
] = (
    weekly_demand[
        "week_period"
    ]
    .dt.end_time
)


weekly_demand = (
    weekly_demand
    .sort_values(
        "week_start"
    )
    .reset_index(drop=True)
)


print(
    f"Observed weekly rows: "
    f"{len(weekly_demand):,}"
)
```

```python
# ================================================================
# 63.10 — COMPLETE WEEKLY CALENDAR
# ================================================================

first_week = (
    demand_lines[
        "week_period"
    ]
    .min()
)


last_week = (
    demand_lines[
        "week_period"
    ]
    .max()
)


complete_week_periods = pd.period_range(
    start=first_week,
    end=last_week,
    freq="W-SUN"
)


complete_weekly_calendar = pd.DataFrame(
    {
        "week_period":
            complete_week_periods
    }
)


complete_weekly_calendar[
    "week_start"
] = (
    complete_weekly_calendar[
        "week_period"
    ]
    .dt.start_time
)


complete_weekly_calendar[
    "week_end"
] = (
    complete_weekly_calendar[
        "week_period"
    ]
    .dt.end_time
)
```

```python
# ================================================================
# 63.11 — COMPLETE WEEKLY SERIES
# ================================================================

weekly_demand = (
    complete_weekly_calendar
    .merge(
        weekly_demand.drop(
            columns=[
                "week_start",
                "week_end",
            ]
        ),
        on="week_period",
        how="left",
        validate="one_to_one"
    )
)
```

```python
# ================================================================
# 63.12 — MISSING WEEK AUDIT
# ================================================================

weekly_demand[
    "is_missing_week"
] = (
    weekly_demand[
        "revenue"
    ]
    .isna()
)


missing_week_count = (
    weekly_demand[
        "is_missing_week"
    ]
    .sum()
)


print(
    f"Missing calendar weeks: "
    f"{missing_week_count:,}"
)


if missing_week_count > 0:

    display(
        weekly_demand.loc[
            weekly_demand[
                "is_missing_week"
            ],
            [
                "week_period",
                "week_start",
                "week_end",
            ]
        ]
    )
```

```python
# ================================================================
# 63.13 — PARTIAL WEEK AUDIT
# ================================================================

data_start_date = (
    demand_lines[
        "date"
    ]
    .min()
)


data_end_date = (
    demand_lines[
        "date"
    ]
    .max()
)


weekly_demand[
    "is_partial_start_week"
] = (
    weekly_demand[
        "week_start"
    ]
    <
    data_start_date.normalize()
)


weekly_demand[
    "is_partial_end_week"
] = (
    weekly_demand[
        "week_end"
    ]
    >
    data_end_date
)


partial_weeks = (
    weekly_demand.loc[
        weekly_demand[
            "is_partial_start_week"
        ]
        |
        weekly_demand[
            "is_partial_end_week"
        ]
    ]
)


display(
    partial_weeks[
        [
            "week_period",
            "week_start",
            "week_end",
            "revenue",
            "orders",
            "units",
            "is_partial_start_week",
            "is_partial_end_week",
        ]
    ]
)
```

```python
# ================================================================
# 63.14 — COMPLETE-WEEK FLAG
# ================================================================

weekly_demand[
    "is_complete_week"
] = (
    ~weekly_demand[
        "is_partial_start_week"
    ]
    &
    ~weekly_demand[
        "is_partial_end_week"
    ]
    &
    ~weekly_demand[
        "is_missing_week"
    ]
)


print(
    "Complete weeks:",
    int(
        weekly_demand[
            "is_complete_week"
        ].sum()
    )
)
```

```python
# ================================================================
# 63.15 — MODELLING WEEKLY SERIES
# ================================================================

weekly_demand_model = (
    weekly_demand.loc[
        weekly_demand[
            "is_complete_week"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


print(
    f"Complete modelling weeks: "
    f"{len(weekly_demand_model):,}"
)

print(
    f"First modelling week: "
    f"{weekly_demand_model['week_start'].min().date()}"
)

print(
    f"Last modelling week: "
    f"{weekly_demand_model['week_end'].max().date()}"
)
```

```python
# ================================================================
# 63.16 — WEEKLY DEMAND SUMMARY
# ================================================================

weekly_demand_summary = (
    weekly_demand_model[
        [
            "revenue",
            "orders",
            "units",
            "active_customers",
            "active_products",
        ]
    ]
    .describe(
        percentiles=[
            0.10,
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99,
        ]
    )
)


display(
    weekly_demand_summary
)
```

```python
# ================================================================
# 63.17 — CALENDAR DIAGNOSTIC FEATURES
# ================================================================

weekly_demand_model[
    "year"
] = (
    weekly_demand_model[
        "week_start"
    ]
    .dt.year
)


weekly_demand_model[
    "month"
] = (
    weekly_demand_model[
        "week_start"
    ]
    .dt.month
)


weekly_demand_model[
    "week_of_year"
] = (
    weekly_demand_model[
        "week_start"
    ]
    .dt.isocalendar()
    .week
    .astype(int)
)


weekly_demand_model[
    "quarter"
] = (
    weekly_demand_model[
        "week_start"
    ]
    .dt.quarter
)
```

```python
# ================================================================
# VISUAL 63.1 — WEEKLY REVENUE OVER TIME
# ================================================================

plt.figure(
    figsize=(14, 6)
)


sns.lineplot(
    data=weekly_demand_model,
    x="week_start",
    y="revenue"
)


plt.title(
    "Weekly Positive Merchandise Revenue"
)

plt.xlabel(
    "Week"
)

plt.ylabel(
    "Revenue"
)

plt.tight_layout()
plt.show()
```

```python
# ================================================================
# VISUAL 63.2 — WEEKLY ORDERS
# ================================================================

plt.figure(
    figsize=(14, 6)
)


sns.lineplot(
    data=weekly_demand_model,
    x="week_start",
    y="orders"
)


plt.title(
    "Weekly Qualifying Orders"
)

plt.xlabel(
    "Week"
)

plt.ylabel(
    "Distinct Orders"
)

plt.tight_layout()
plt.show()
```

```python
# ================================================================
# VISUAL 63.3 — WEEKLY UNITS
# ================================================================

plt.figure(
    figsize=(14, 6)
)


sns.lineplot(
    data=weekly_demand_model,
    x="week_start",
    y="units"
)


plt.title(
    "Weekly Merchandise Units"
)

plt.xlabel(
    "Week"
)

plt.ylabel(
    "Units"
)

plt.tight_layout()
plt.show()
```

```python
# ================================================================
# 63.21 — 4-WEEK ROLLING DIAGNOSTICS
# ================================================================

weekly_demand_model[
    "revenue_4w_ma"
] = (
    weekly_demand_model[
        "revenue"
    ]
    .rolling(
        window=4,
        min_periods=1
    )
    .mean()
)


weekly_demand_model[
    "orders_4w_ma"
] = (
    weekly_demand_model[
        "orders"
    ]
    .rolling(
        window=4,
        min_periods=1
    )
    .mean()
)
```

```python
# ================================================================
# VISUAL 63.4 — REVENUE WITH 4-WEEK MOVING AVERAGE
# ================================================================

rolling_revenue_visual = (
    weekly_demand_model[
        [
            "week_start",
            "revenue",
            "revenue_4w_ma",
        ]
    ]
    .melt(
        id_vars="week_start",
        var_name="Series",
        value_name="Revenue"
    )
)


plt.figure(
    figsize=(14, 6)
)


sns.lineplot(
    data=rolling_revenue_visual,
    x="week_start",
    y="Revenue",
    hue="Series"
)


plt.title(
    "Weekly Revenue and 4-Week Moving Average"
)

plt.xlabel(
    "Week"
)

plt.ylabel(
    "Revenue"
)

plt.tight_layout()
plt.show()
```

```python
# ================================================================
# 63.22 — WEEK-OVER-WEEK REVENUE CHANGE
# ================================================================

weekly_demand_model[
    "revenue_wow_change"
] = (
    weekly_demand_model[
        "revenue"
    ]
    .pct_change()
)


weekly_demand_model[
    "orders_wow_change"
] = (
    weekly_demand_model[
        "orders"
    ]
    .pct_change()
)
```

```python
weekly_movement_summary = (
    weekly_demand_model[
        [
            "revenue_wow_change",
            "orders_wow_change",
        ]
    ]
    .describe(
        percentiles=[
            0.05,
            0.10,
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
        ]
    )
)


display(
    weekly_movement_summary
)
```

```python
# ================================================================
# 63.23 — REVENUE VOLATILITY
# ================================================================

weekly_revenue_mean = (
    weekly_demand_model[
        "revenue"
    ]
    .mean()
)


weekly_revenue_std = (
    weekly_demand_model[
        "revenue"
    ]
    .std()
)


weekly_revenue_cv = (
    weekly_revenue_std
    /
    weekly_revenue_mean
)


print(
    f"Weekly revenue mean: "
    f"{weekly_revenue_mean:,.2f}"
)

print(
    f"Weekly revenue std: "
    f"{weekly_revenue_std:,.2f}"
)

print(
    f"Weekly revenue coefficient of variation: "
    f"{weekly_revenue_cv:.3f}"
)
```

```python
# ================================================================
# 63.24 — DEMAND DRIVER CORRELATIONS
# ================================================================

weekly_demand_correlations = (
    weekly_demand_model[
        [
            "revenue",
            "orders",
            "units",
            "active_customers",
            "active_products",
        ]
    ]
    .corr(
        method="spearman"
    )
)


display(
    weekly_demand_correlations.round(3)
)
```

```python
# ================================================================
# VISUAL 63.5 — DEMAND DRIVER CORRELATIONS
# ================================================================

plt.figure(
    figsize=(8, 6)
)


sns.heatmap(
    weekly_demand_correlations,
    annot=True,
    fmt=".2f"
)


plt.title(
    "Weekly Demand Signal Correlations"
)

plt.tight_layout()
plt.show()
```

```python
# ================================================================
# 63.25 — MONTH-OF-YEAR DEMAND PROFILE
# ================================================================

monthly_demand_profile = (
    weekly_demand_model
    .groupby(
        "month",
        as_index=False
    )
    .agg(
        weeks=(
            "revenue",
            "size"
        ),

        average_weekly_revenue=(
            "revenue",
            "mean"
        ),

        median_weekly_revenue=(
            "revenue",
            "median"
        ),

        average_weekly_orders=(
            "orders",
            "mean"
        ),

        average_weekly_units=(
            "units",
            "mean"
        ),
    )
)


display(
    monthly_demand_profile
)
```

```python
plt.figure(
    figsize=(10, 6)
)


sns.barplot(
    data=monthly_demand_profile,
    x="month",
    y="average_weekly_revenue"
)


plt.title(
    "Average Weekly Revenue by Calendar Month"
)

plt.xlabel(
    "Month"
)

plt.ylabel(
    "Average Weekly Revenue"
)

plt.tight_layout()
plt.show()
```

```python
# ================================================================
# 63.27 — EXTREME DEMAND WEEKS
# ================================================================

highest_revenue_weeks = (
    weekly_demand_model
    .nlargest(
        10,
        "revenue"
    )[
        [
            "week_start",
            "week_end",
            "revenue",
            "orders",
            "units",
            "active_customers",
            "active_products",
        ]
    ]
)


lowest_revenue_weeks = (
    weekly_demand_model
    .nsmallest(
        10,
        "revenue"
    )[
        [
            "week_start",
            "week_end",
            "revenue",
            "orders",
            "units",
            "active_customers",
            "active_products",
        ]
    ]
)


print(
    "Highest revenue weeks:"
)

display(
    highest_revenue_weeks
)


print(
    "Lowest revenue weeks:"
)

display(
    lowest_revenue_weeks
)
```

## Step 64 — Temporal Split and Forecasting Baselines

Step 64 establishes the baseline forecasting performance that all later models must beat.

The weekly demand series is now regular and includes observed year-end zero-trading weeks.

The primary forecasting target remains:

> **weekly positive merchandise revenue**

Four simple forecasting approaches will initially be evaluated:

### Last-Week Naive

> Forecast next week's revenue using the most recently observed week.

### Four-Week Moving Average

> Forecast using the average revenue of the preceding four weeks.

### Eight-Week Moving Average

> Forecast using the average revenue of the preceding eight weeks.

### Seasonal Naive — 52 Weeks

> Forecast using revenue observed exactly 52 weeks earlier.

The seasonal naive model is especially important because Step 63 identified strong annual recurrence.

All predictions are generated using historical information available before the forecast week.

The final Test period will remain unopened until the forecasting architecture is selected and frozen.

```python
# ================================================================
# 64.2 — CALENDAR FEATURES
# ================================================================

weekly_demand_model[
    "year"
] = (
    weekly_demand_model[
        "week_start"
    ]
    .dt.year
)


weekly_demand_model[
    "month"
] = (
    weekly_demand_model[
        "week_start"
    ]
    .dt.month
)


weekly_demand_model[
    "week_of_year"
] = (
    weekly_demand_model[
        "week_start"
    ]
    .dt.isocalendar()
    .week
    .astype(int)
)
```

```python
# ================================================================
# 64.3 — BASELINE FORECAST FEATURES
# ================================================================

weekly_demand_model[
    "forecast_naive_1w"
] = (
    weekly_demand_model[
        "revenue"
    ]
    .shift(1)
)


weekly_demand_model[
    "forecast_ma4"
] = (
    weekly_demand_model[
        "revenue"
    ]
    .shift(1)
    .rolling(
        window=4,
        min_periods=4
    )
    .mean()
)


weekly_demand_model[
    "forecast_ma8"
] = (
    weekly_demand_model[
        "revenue"
    ]
    .shift(1)
    .rolling(
        window=8,
        min_periods=8
    )
    .mean()
)


weekly_demand_model[
    "forecast_seasonal_52"
] = (
    weekly_demand_model[
        "revenue"
    ]
    .shift(52)
)
```

```python
# ================================================================
# 64.4 — SIMPLE SEASONAL / RECENT BLEND
# ================================================================

weekly_demand_model[
    "forecast_blend_52_ma4"
] = (
    0.50
    *
    weekly_demand_model[
        "forecast_seasonal_52"
    ]
    +
    0.50
    *
    weekly_demand_model[
        "forecast_ma4"
    ]
)
```

```python
# ================================================================
# 64.5 — FORECAST METRICS
# ================================================================

def evaluate_forecast(
    actual,
    predicted,
    model_name
):

    actual = np.asarray(
        actual,
        dtype=float
    )


    predicted = np.asarray(
        predicted,
        dtype=float
    )


    valid = (
        np.isfinite(
            actual
        )
        &
        np.isfinite(
            predicted
        )
    )


    actual = actual[
        valid
    ]

    predicted = predicted[
        valid
    ]


    errors = (
        predicted
        -
        actual
    )


    absolute_errors = np.abs(
        errors
    )


    mae = (
        absolute_errors.mean()
    )


    rmse = np.sqrt(
        np.mean(
            errors ** 2
        )
    )


    wape = (
        absolute_errors.sum()
        /
        actual.sum()
        if actual.sum() != 0
        else np.nan
    )


    bias = (
        errors.mean()
    )


    total_ratio = (
        predicted.sum()
        /
        actual.sum()
        if actual.sum() != 0
        else np.nan
    )


    return pd.DataFrame(
        {
            "Model": [
                model_name
            ],

            "Observations": [
                len(
                    actual
                )
            ],

            "MAE": [
                mae
            ],

            "RMSE": [
                rmse
            ],

            "WAPE": [
                wape
            ],

            "Bias": [
                bias
            ],

            "Predicted / Actual": [
                total_ratio
            ],
        }
    )
```

```python
# ================================================================
# 64.6 — COMMON BASELINE EVALUATION WINDOW
# ================================================================

baseline_forecast_columns = [
    "forecast_naive_1w",
    "forecast_ma4",
    "forecast_ma8",
    "forecast_seasonal_52",
    "forecast_blend_52_ma4",
]


baseline_common_evaluation = (
    weekly_demand_model.dropna(
        subset=
            baseline_forecast_columns
    )
    .copy()
    .reset_index(drop=True)
)


print(
    f"Common baseline observations: "
    f"{len(baseline_common_evaluation):,}"
)

print(
    f"Evaluation start: "
    f"{baseline_common_evaluation['week_start'].min().date()}"
)

print(
    f"Evaluation end: "
    f"{baseline_common_evaluation['week_end'].max().date()}"
)
```

```python
# ================================================================
# 64.7 — BASELINE COMPARISON
# ================================================================

baseline_results = []


baseline_model_map = {
    "Last-Week Naive":
        "forecast_naive_1w",

    "4-Week Moving Average":
        "forecast_ma4",

    "8-Week Moving Average":
        "forecast_ma8",

    "Seasonal Naive — 52 Weeks":
        "forecast_seasonal_52",

    "50/50 Seasonal + MA4":
        "forecast_blend_52_ma4",
}


for (
    model_name,
    forecast_column
) in (
    baseline_model_map.items()
):

    result = (
        evaluate_forecast(
            actual=
                baseline_common_evaluation[
                    "revenue"
                ],

            predicted=
                baseline_common_evaluation[
                    forecast_column
                ],

            model_name=
                model_name
        )
    )


    baseline_results.append(
        result
    )


forecast_baseline_results = (
    pd.concat(
        baseline_results,
        ignore_index=True
    )
    .sort_values(
        "WAPE"
    )
    .reset_index(drop=True)
)


display(
    forecast_baseline_results
)
```

```python
# ================================================================
# 64.12 — WEEKLY BASELINE WINNERS
# ================================================================

winner_records = []


for row in (
    baseline_common_evaluation
    .itertuples(
        index=False
    )
):

    forecasts = {
        "Last-Week Naive":
            row.forecast_naive_1w,

        "4-Week Moving Average":
            row.forecast_ma4,

        "8-Week Moving Average":
            row.forecast_ma8,

        "Seasonal Naive — 52 Weeks":
            row.forecast_seasonal_52,

        "50/50 Seasonal + MA4":
            row.forecast_blend_52_ma4,
    }


    errors = {
        model:
            abs(
                prediction
                -
                row.revenue
            )

        for (
            model,
            prediction
        ) in forecasts.items()
    }


    winner = min(
        errors,
        key=
            errors.get
    )


    winner_records.append(
        {
            "week_start":
                row.week_start,

            "actual_revenue":
                row.revenue,

            "best_baseline":
                winner,

            "winning_absolute_error":
                errors[
                    winner
                ],
        }
    )


weekly_baseline_winners = pd.DataFrame(
    winner_records
)


baseline_winner_summary = (
    weekly_baseline_winners[
        "best_baseline"
    ]
    .value_counts()
    .rename_axis(
        "Model"
    )
    .reset_index(
        name="Winning Weeks"
    )
)


baseline_winner_summary[
    "Winning Share"
] = (
    baseline_winner_summary[
        "Winning Weeks"
    ]
    /
    len(
        weekly_baseline_winners
    )
)


display(
    baseline_winner_summary
)
```

```python
# ================================================================
# 65.2 — FORECAST VALIDATION GOVERNANCE
# ================================================================

FORECAST_FINAL_HOLDOUT_PRISTINE = False

FORECAST_VALIDATION_METHOD = (
    "Expanding-window rolling-origin backtesting"
)

FORECAST_MODEL_FROZEN = False


print(
    f"Validation method: "
    f"{FORECAST_VALIDATION_METHOD}"
)

print(
    "Pristine final weekly holdout available:",
    FORECAST_FINAL_HOLDOUT_PRISTINE
)
```

```python
# ================================================================
# 65.3 — FORECAST FEATURE ENGINEERING
# ================================================================

forecast_data = (
    weekly_demand_model.copy()
)


# ------------------------------------------------
# Revenue lags
# ------------------------------------------------

for lag in [
    1,
    2,
    3,
    4,
    8,
    13,
    26,
    52,
]:

    forecast_data[
        f"revenue_lag_{lag}"
    ] = (
        forecast_data[
            "revenue"
        ]
        .shift(
            lag
        )
    )


# ------------------------------------------------
# Leakage-safe rolling revenue
# ------------------------------------------------

for window in [
    4,
    8,
    13,
]:

    forecast_data[
        f"revenue_ma_{window}"
    ] = (
        forecast_data[
            "revenue"
        ]
        .shift(1)
        .rolling(
            window=window,
            min_periods=window
        )
        .mean()
    )


# ------------------------------------------------
# Recent volatility
# ------------------------------------------------

forecast_data[
    "revenue_std_4"
] = (
    forecast_data[
        "revenue"
    ]
    .shift(1)
    .rolling(
        window=4,
        min_periods=4
    )
    .std()
)


forecast_data[
    "revenue_std_8"
] = (
    forecast_data[
        "revenue"
    ]
    .shift(1)
    .rolling(
        window=8,
        min_periods=8
    )
    .std()
)
```

```python
# ================================================================
# 65.4 — SUPPORTING SIGNAL LAGS
# ================================================================

for signal in [
    "orders",
    "units",
    "active_customers",
    "active_products",
]:

    forecast_data[
        f"{signal}_lag_1"
    ] = (
        forecast_data[
            signal
        ]
        .shift(1)
    )


    forecast_data[
        f"{signal}_ma_4"
    ] = (
        forecast_data[
            signal
        ]
        .shift(1)
        .rolling(
            window=4,
            min_periods=4
        )
        .mean()
    )
```

```python
# ================================================================
# 65.5 — CALENDAR FEATURES
# ================================================================

forecast_data[
    "week_of_year"
] = (
    forecast_data[
        "week_start"
    ]
    .dt.isocalendar()
    .week
    .astype(int)
)


forecast_data[
    "week_sin"
] = np.sin(
    2
    *
    np.pi
    *
    forecast_data[
        "week_of_year"
    ]
    /
    52
)


forecast_data[
    "week_cos"
] = np.cos(
    2
    *
    np.pi
    *
    forecast_data[
        "week_of_year"
    ]
    /
    52
)


forecast_data[
    "time_index"
] = np.arange(
    len(
        forecast_data
    )
)
```

```python
# ================================================================
# 65.6 — YEAR-END CALENDAR INDICATOR
# ================================================================

forecast_data[
    "year_end_shutdown_week"
] = (
    forecast_data[
        "week_of_year"
    ]
    .isin(
        [
            52,
            53,
        ]
    )
    .astype(int)
)
```

```python
# ================================================================
# 65.7 — MODEL FEATURE SET
# ================================================================

forecast_feature_columns = [
    "revenue_lag_1",
    "revenue_lag_2",
    "revenue_lag_3",
    "revenue_lag_4",
    "revenue_lag_8",
    "revenue_lag_13",
    "revenue_lag_26",
    "revenue_lag_52",

    "revenue_ma_4",
    "revenue_ma_8",
    "revenue_ma_13",

    "revenue_std_4",
    "revenue_std_8",

    "orders_lag_1",
    "orders_ma_4",

    "units_lag_1",
    "units_ma_4",

    "active_customers_lag_1",
    "active_customers_ma_4",

    "active_products_lag_1",
    "active_products_ma_4",

    "week_sin",
    "week_cos",
    "time_index",
    "year_end_shutdown_week",
]
```

```python
# ================================================================
# 65.8 — COMPLETE FEATURE POPULATION
# ================================================================

forecast_model_data = (
    forecast_data.dropna(
        subset=
            forecast_feature_columns
            +
            [
                "revenue"
            ]
    )
    .copy()
    .reset_index(drop=True)
)


print(
    f"Forecast modelling rows: "
    f"{len(forecast_model_data):,}"
)

print(
    f"First eligible forecast week: "
    f"{forecast_model_data['week_start'].min().date()}"
)

print(
    f"Last eligible forecast week: "
    f"{forecast_model_data['week_end'].max().date()}"
)
```

```python
# ================================================================
# 65.9 — ROLLING-ORIGIN DESIGN
# ================================================================

MIN_ROLLING_TRAIN_WEEKS = 12


rolling_origins = list(
    range(
        MIN_ROLLING_TRAIN_WEEKS,
        len(
            forecast_model_data
        )
    )
)


print(
    f"Rolling-origin forecast evaluations: "
    f"{len(rolling_origins):,}"
)
```

```python
# ================================================================
# 65.10 — MODEL IMPORTS
# ================================================================

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import Ridge

from sklearn.ensemble import GradientBoostingRegressor
```

```python
# ================================================================
# 65.11 — RIDGE FORECAST MODEL
# ================================================================

ridge_forecast_model = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        ),

        (
            "model",
            Ridge(
                alpha=10.0
            )
        ),
    ]
)
```

```python
# ================================================================
# 65.12 — GRADIENT BOOSTING FORECAST MODEL
# ================================================================

gb_forecast_model = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "model",
            GradientBoostingRegressor(
                n_estimators=100,
                learning_rate=0.03,
                max_depth=2,
                min_samples_leaf=5,
                loss="huber",
                random_state=42
            )
        ),
    ]
)
```

```python
# ================================================================
# 65.13 — BLEND CANDIDATES
# ================================================================

seasonal_weights = [
    0.50,
    0.60,
    0.70,
    0.80,
    0.90,
    1.00,
]


print(
    "Seasonal blend weights:",
    seasonal_weights
)
```

```python
# ================================================================
# 65.14 — EXPANDING-WINDOW BACKTEST
# ================================================================

from sklearn.base import clone


rolling_forecast_records = []


for origin_index in (
    rolling_origins
):

    train = (
        forecast_model_data
        .iloc[
            :origin_index
        ]
        .copy()
    )


    validation = (
        forecast_model_data
        .iloc[
            [
                origin_index
            ]
        ]
        .copy()
    )


    X_train = (
        train[
            forecast_feature_columns
        ]
    )


    y_train = (
        train[
            "revenue"
        ]
    )


    X_validation = (
        validation[
            forecast_feature_columns
        ]
    )


    actual = (
        validation[
            "revenue"
        ]
        .iloc[0]
    )


    # ---------------------------------------------
    # Seasonal Naive
    # ---------------------------------------------

    seasonal_prediction = (
        validation[
            "revenue_lag_52"
        ]
        .iloc[0]
    )


    rolling_forecast_records.append(
        {
            "week_start":
                validation[
                    "week_start"
                ].iloc[0],

            "model":
                "Seasonal Naive — 52 Weeks",

            "actual":
                actual,

            "prediction":
                seasonal_prediction,
        }
    )


    # ---------------------------------------------
    # Blend candidates
    # ---------------------------------------------

    for seasonal_weight in (
        seasonal_weights
    ):

        blend_prediction = (
            seasonal_weight
            *
            validation[
                "revenue_lag_52"
            ].iloc[0]
            +
            (
                1
                -
                seasonal_weight
            )
            *
            validation[
                "revenue_ma_4"
            ].iloc[0]
        )


        rolling_forecast_records.append(
            {
                "week_start":
                    validation[
                        "week_start"
                    ].iloc[0],

                "model":
                    (
                        f"Seasonal Blend "
                        f"{seasonal_weight:.1f}"
                    ),

                "actual":
                    actual,

                "prediction":
                    blend_prediction,
            }
        )


    # ---------------------------------------------
    # Ridge
    # ---------------------------------------------

    ridge_model = clone(
        ridge_forecast_model
    )


    ridge_model.fit(
        X_train,
        y_train
    )


    ridge_prediction = (
        ridge_model.predict(
            X_validation
        )[0]
    )


    rolling_forecast_records.append(
        {
            "week_start":
                validation[
                    "week_start"
                ].iloc[0],

            "model":
                "Ridge Lag Forecast",

            "actual":
                actual,

            "prediction":
                max(
                    0,
                    ridge_prediction
                ),
        }
    )


    # ---------------------------------------------
    # Gradient Boosting
    # ---------------------------------------------

    gb_model = clone(
        gb_forecast_model
    )


    gb_model.fit(
        X_train,
        y_train
    )


    gb_prediction = (
        gb_model.predict(
            X_validation
        )[0]
    )


    rolling_forecast_records.append(
        {
            "week_start":
                validation[
                    "week_start"
                ].iloc[0],

            "model":
                "Gradient Boosting Lag Forecast",

            "actual":
                actual,

            "prediction":
                max(
                    0,
                    gb_prediction
                ),
        }
    )


rolling_forecasts = pd.DataFrame(
    rolling_forecast_records
)


print(
    f"Rolling predictions created: "
    f"{len(rolling_forecasts):,}"
)
```

```python
# ================================================================
# 65.15 — ROLLING FORECAST PERFORMANCE
# ================================================================

rolling_forecast_results = []


for model_name, group in (
    rolling_forecasts
    .groupby(
        "model"
    )
):

    result = (
        evaluate_forecast(
            actual=
                group[
                    "actual"
                ],

            predicted=
                group[
                    "prediction"
                ],

            model_name=
                model_name
        )
    )


    rolling_forecast_results.append(
        result
    )


rolling_forecast_summary = (
    pd.concat(
        rolling_forecast_results,
        ignore_index=True
    )
    .sort_values(
        "WAPE"
    )
    .reset_index(drop=True)
)


display(
    rolling_forecast_summary
)
```

```python
# ================================================================
# 65.17 — MONTHLY FORECAST STABILITY
# ================================================================

rolling_forecasts[
    "month"
] = (
    rolling_forecasts[
        "week_start"
    ]
    .dt.month
)


rolling_monthly_performance_records = []


for (
    model_name,
    month
), group in (
    rolling_forecasts
    .groupby(
        [
            "model",
            "month",
        ]
    )
):

    if len(
        group
    ) < 2:

        continue


    result = (
        evaluate_forecast(
            actual=
                group[
                    "actual"
                ],

            predicted=
                group[
                    "prediction"
                ],

            model_name=
                model_name
        )
    )


    record = (
        result.iloc[0]
        .to_dict()
    )


    record[
        "Month"
    ] = month


    rolling_monthly_performance_records.append(
        record
    )


rolling_monthly_performance = pd.DataFrame(
    rolling_monthly_performance_records
)


display(
    rolling_monthly_performance
)
```

## Step 66 — Multi-Horizon Forecast Robustness

Step 66 evaluates the surviving forecasting architecture across the operational planning horizons defined at the start of Notebook 7.

- **2 weeks ahead**
- **4 weeks ahead**

```python
# ================================================================
# 66.2 — SURVIVING BLEND WEIGHTS
# ================================================================

MULTIHORIZON_SEASONAL_WEIGHTS = [
    0.50,
    0.60,
    0.70,
    0.80,
    0.90,
    1.00,
]


FORECAST_HORIZONS = [
    1,
    2,
    4,
]


print(
    "Forecast horizons:",
    FORECAST_HORIZONS
)

print(
    "Seasonal weights:",
    MULTIHORIZON_SEASONAL_WEIGHTS
)
```

```python
# ================================================================
# 66.3 — ORIGIN-LEVEL FORECAST DATA
# ================================================================

multihorizon_data = (
    weekly_demand_model[
        [
            "week_start",
            "week_end",
            "revenue",
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


multihorizon_data[
    "recent_ma4_at_origin"
] = (
    multihorizon_data[
        "revenue"
    ]
    .rolling(
        window=4,
        min_periods=4
    )
    .mean()
)
```

### 66.4 — Multi-horizon forecast generator
For horizon h:
- target = revenue h weeks after the origin;
- annual seasonal analogue = target week minus 52 weeks;
- recent component = the four-week average known at the origin.

```python
# ================================================================
# 66.4 — MULTI-HORIZON FORECAST GENERATOR
# ================================================================

multihorizon_records = []


for origin_index in range(
    len(multihorizon_data)
):

    origin_row = (
        multihorizon_data
        .iloc[
            origin_index
        ]
    )


    recent_ma4 = (
        origin_row[
            "recent_ma4_at_origin"
        ]
    )


    if pd.isna(
        recent_ma4
    ):
        continue


    for horizon in (
        FORECAST_HORIZONS
    ):

        target_index = (
            origin_index
            +
            horizon
        )


        if (
            target_index
            >=
            len(
                multihorizon_data
            )
        ):
            continue


        seasonal_index = (
            target_index
            -
            52
        )


        if (
            seasonal_index
            <
            0
        ):
            continue


        target_row = (
            multihorizon_data
            .iloc[
                target_index
            ]
        )


        actual = (
            target_row[
                "revenue"
            ]
        )


        seasonal_prediction = (
            multihorizon_data
            .iloc[
                seasonal_index
            ][
                "revenue"
            ]
        )


        # ---------------------------------------------
        # Pure seasonal benchmark
        # ---------------------------------------------

        multihorizon_records.append(
            {
                "origin_week":
                    origin_row[
                        "week_start"
                    ],

                "target_week":
                    target_row[
                        "week_start"
                    ],

                "horizon":
                    horizon,

                "model":
                    "Seasonal Naive — 52 Weeks",

                "seasonal_weight":
                    1.00,

                "actual":
                    actual,

                "prediction":
                    seasonal_prediction,
            }
        )


        # ---------------------------------------------
        # Blend candidates
        # ---------------------------------------------

        for seasonal_weight in (
            MULTIHORIZON_SEASONAL_WEIGHTS
        ):

            if (
                seasonal_weight
                ==
                1.00
            ):
                continue


            prediction = (
                seasonal_weight
                *
                seasonal_prediction
                +
                (
                    1
                    -
                    seasonal_weight
                )
                *
                recent_ma4
            )


            multihorizon_records.append(
                {
                    "origin_week":
                        origin_row[
                            "week_start"
                        ],

                    "target_week":
                        target_row[
                            "week_start"
                        ],

                    "horizon":
                        horizon,

                    "model":
                        (
                            f"Seasonal Blend "
                            f"{seasonal_weight:.1f}"
                        ),

                    "seasonal_weight":
                        seasonal_weight,

                    "actual":
                        actual,

                    "prediction":
                        prediction,
                }
            )


multihorizon_forecasts = pd.DataFrame(
    multihorizon_records
)


print(
    f"Multi-horizon forecast records: "
    f"{len(multihorizon_forecasts):,}"
)
```

### 66.11 — Aggregate 4-week planning-block evaluation
A planner may care more about the total demand over the next four weeks than whether each individual week's prediction is exact.

This is a distinct metric and worth testing.

For each forecast origin, we'll compare:
> actual total revenue over next 4 weeks

against
> sum of the four forecasts made at that same origin.

```python
# ================================================================
# 66.11 — 4-WEEK CUMULATIVE PLANNING FORECAST
# ================================================================

four_week_planning_records = []


eligible_origins = (
    multihorizon_data.index[
        multihorizon_data[
            "recent_ma4_at_origin"
        ]
        .notna()
    ]
)


for origin_index in (
    eligible_origins
):

    target_indices = [
        origin_index + h
        for h in [
            1,
            2,
            3,
            4,
        ]
    ]


    if (
        max(
            target_indices
        )
        >=
        len(
            multihorizon_data
        )
    ):
        continue


    seasonal_indices = [
        target_index - 52
        for target_index
        in target_indices
    ]


    if (
        min(
            seasonal_indices
        )
        <
        0
    ):
        continue


    actual_total = (
        multihorizon_data
        .iloc[
            target_indices
        ][
            "revenue"
        ]
        .sum()
    )


    recent_ma4 = (
        multihorizon_data
        .iloc[
            origin_index
        ][
            "recent_ma4_at_origin"
        ]
    )


    seasonal_values = (
        multihorizon_data
        .iloc[
            seasonal_indices
        ][
            "revenue"
        ]
        .to_numpy()
    )


    # Seasonal naive
    four_week_planning_records.append(
        {
            "origin_week":
                multihorizon_data
                .iloc[
                    origin_index
                ][
                    "week_start"
                ],

            "model":
                "Seasonal Naive — 52 Weeks",

            "actual":
                actual_total,

            "prediction":
                seasonal_values.sum(),
        }
    )


    # Blends
    for seasonal_weight in (
        MULTIHORIZON_SEASONAL_WEIGHTS
    ):

        if (
            seasonal_weight
            ==
            1.00
        ):
            continue


        weekly_predictions = (
            seasonal_weight
            *
            seasonal_values
            +
            (
                1
                -
                seasonal_weight
            )
            *
            recent_ma4
        )


        four_week_planning_records.append(
            {
                "origin_week":
                    multihorizon_data
                    .iloc[
                        origin_index
                    ][
                        "week_start"
                    ],

                "model":
                    (
                        f"Seasonal Blend "
                        f"{seasonal_weight:.1f}"
                    ),

                "actual":
                    actual_total,

                "prediction":
                    weekly_predictions.sum(),
            }
        )


four_week_planning_forecasts = (
    pd.DataFrame(
        four_week_planning_records
    )
)
```

### Step 67.5 — Empirical error quantiles
We'll create two planning ranges:

> Central 80% historical range
>>10th–90th percentile error.

>Central 90% historical range
>>5th–95th percentile error.

```python
# ================================================================
# 67.5 — EMPIRICAL INTERVAL QUANTILES
# ================================================================

forecast_error_quantiles = (
    frozen_forecast_backtest
    .groupby(
        "horizon"
    )[
        "error"
    ]
    .quantile(
        [
            0.05,
            0.10,
            0.50,
            0.90,
            0.95,
        ]
    )
    .unstack()
    .reset_index()
)


forecast_error_quantiles.columns = [
    "Horizon",
    "Error P05",
    "Error P10",
    "Error Median",
    "Error P90",
    "Error P95",
]


display(
    forecast_error_quantiles
)
```

```python
# ================================================================
# 67.6 — HISTORICAL PLANNING INTERVALS
# ================================================================

frozen_interval_backtest = (
    frozen_forecast_backtest
    .merge(
        forecast_error_quantiles,
        left_on=
            "horizon",
        right_on=
            "Horizon",
        how=
            "left",
        validate=
            "many_to_one"
    )
)


# Central 80% interval

frozen_interval_backtest[
    "lower_80"
] = np.maximum(
    0,

    frozen_interval_backtest[
        "prediction"
    ]
    +
    frozen_interval_backtest[
        "Error P10"
    ]
)


frozen_interval_backtest[
    "upper_80"
] = np.maximum(
    0,

    frozen_interval_backtest[
        "prediction"
    ]
    +
    frozen_interval_backtest[
        "Error P90"
    ]
)


# Central 90% interval

frozen_interval_backtest[
    "lower_90"
] = np.maximum(
    0,

    frozen_interval_backtest[
        "prediction"
    ]
    +
    frozen_interval_backtest[
        "Error P05"
    ]
)


frozen_interval_backtest[
    "upper_90"
] = np.maximum(
    0,

    frozen_interval_backtest[
        "prediction"
    ]
    +
    frozen_interval_backtest[
        "Error P95"
    ]
)
```

```python
# ================================================================
# 67.7 — INTERVAL COVERAGE
# ================================================================

frozen_interval_backtest[
    "covered_80"
] = (
    (
        frozen_interval_backtest[
            "actual"
        ]
        >=
        frozen_interval_backtest[
            "lower_80"
        ]
    )
    &
    (
        frozen_interval_backtest[
            "actual"
        ]
        <=
        frozen_interval_backtest[
            "upper_80"
        ]
    )
)


frozen_interval_backtest[
    "covered_90"
] = (
    (
        frozen_interval_backtest[
            "actual"
        ]
        >=
        frozen_interval_backtest[
            "lower_90"
        ]
    )
    &
    (
        frozen_interval_backtest[
            "actual"
        ]
        <=
        frozen_interval_backtest[
            "upper_90"
        ]
    )
)


forecast_interval_coverage = (
    frozen_interval_backtest
    .groupby(
        "horizon",
        as_index=False
    )
    .agg(
        observations=(
            "actual",
            "size"
        ),

        coverage_80=(
            "covered_80",
            "mean"
        ),

        coverage_90=(
            "covered_90",
            "mean"
        ),

        average_80_width=(
            "upper_80",
            lambda values:
                np.nan
        ),
    )
)
```

```python
# ================================================================
# 67.7B — INTERVAL WIDTH
# ================================================================

frozen_interval_backtest[
    "width_80"
] = (
    frozen_interval_backtest[
        "upper_80"
    ]
    -
    frozen_interval_backtest[
        "lower_80"
    ]
)


frozen_interval_backtest[
    "width_90"
] = (
    frozen_interval_backtest[
        "upper_90"
    ]
    -
    frozen_interval_backtest[
        "lower_90"
    ]
)


forecast_interval_coverage = (
    frozen_interval_backtest
    .groupby(
        "horizon",
        as_index=False
    )
    .agg(
        observations=(
            "actual",
            "size"
        ),

        coverage_80=(
            "covered_80",
            "mean"
        ),

        coverage_90=(
            "covered_90",
            "mean"
        ),

        average_80_width=(
            "width_80",
            "mean"
        ),

        average_90_width=(
            "width_90",
            "mean"
        ),
    )
)


display(
    forecast_interval_coverage
)
```

## Step 68 — Product-Level Demand Feasibility and Segmentation

Step 68 evaluates whether individual products contain sufficient weekly demand history to support standalone forecasting.

The catalogue is expected to contain a substantial long tail.

For many products, weekly demand may be:

- intermittent;
- sparse;
- highly volatile;
- newly introduced;
- or observed for too few weeks.

Forecasting every product independently would therefore create false precision.

This step segments products according to historical demand structure before any product-level forecasting model is selected.

The analysis focuses primarily on:

> **weekly positive merchandise units**

Revenue will remain available as a supporting product-level signal.

```python
# ================================================================
# 68.2 — WEEKLY PRODUCT DEMAND
# ================================================================

product_weekly_observed = (
    demand_lines
    .groupby(
        [
            "product",
            "week_period",
        ],
        as_index=False
    )
    .agg(
        units=(
            "quantity",
            "sum"
        ),

        revenue=(
            "net_merchandise_revenue",
            "sum"
        ),

        orders=(
            "invoice",
            "nunique"
        ),

        customers=(
            "customer",
            lambda values:
                values.dropna().nunique()
        ),
    )
)


print(
    f"Observed product-week rows: "
    f"{len(product_weekly_observed):,}"
)

print(
    f"Distinct products: "
    f"{product_weekly_observed['product'].nunique():,}"
)
```

```python
# ================================================================
# 68.3 — PRODUCT FORECASTING CALENDAR
# ================================================================

product_forecast_calendar = (
    weekly_demand_model[
        [
            "week_period",
            "week_start",
            "week_end",
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


print(
    f"Product forecasting weeks: "
    f"{len(product_forecast_calendar):,}"
)
```

```python
# ================================================================
# 68.4 — PRODUCT FIRST-OBSERVED WEEK
# ================================================================

product_first_observed = (
    product_weekly_observed
    .groupby(
        "product",
        as_index=False
    )
    .agg(
        first_observed_week=(
            "week_period",
            "min"
        ),

        last_observed_week=(
            "week_period",
            "max"
        ),
    )
)


display(
    product_first_observed.head()
)
```

### 68.5 — Build post-entry product-week panel
For each product, weeks after first observation count as either:
- positive demand;
- or zero observed demand.

Weeks before first appearance are left-truncated and should not be treated as zero.

```python
# ================================================================
# 68.5 — BUILD POST-ENTRY PRODUCT-WEEK PANEL
# ================================================================

product_week_panels = []


for row in (
    product_first_observed
    .itertuples(
        index=False
    )
):

    calendar_subset = (
        product_forecast_calendar.loc[
            product_forecast_calendar[
                "week_period"
            ]
            >=
            row.first_observed_week
        ]
        .copy()
    )


    calendar_subset[
        "product"
    ] = row.product


    product_week_panels.append(
        calendar_subset
    )


product_week_panel = (
    pd.concat(
        product_week_panels,
        ignore_index=True
    )
)


product_week_panel = (
    product_week_panel
    .merge(
        product_weekly_observed,
        on=[
            "product",
            "week_period",
        ],
        how="left",
        validate="one_to_one"
    )
)


for column in [
    "units",
    "revenue",
    "orders",
    "customers",
]:

    product_week_panel[
        column
    ] = (
        product_week_panel[
            column
        ]
        .fillna(0)
    )


print(
    f"Product-week panel rows: "
    f"{len(product_week_panel):,}"
)
```

```python
# ================================================================
# 68.6 — PRODUCT DEMAND PROFILE
# ================================================================

def positive_week_cv(values):

    positive = (
        values[
            values > 0
        ]
    )


    if (
        len(
            positive
        )
        <
        2
    ):
        return np.nan


    mean_value = (
        positive.mean()
    )


    if (
        mean_value
        ==
        0
    ):
        return np.nan


    return (
        positive.std()
        /
        mean_value
    )


product_demand_profile = (
    product_week_panel
    .groupby(
        "product",
        as_index=False
    )
    .agg(
        observed_weeks=(
            "units",
            "size"
        ),

        positive_demand_weeks=(
            "units",
            lambda values:
                (values > 0).sum()
        ),

        zero_demand_weeks=(
            "units",
            lambda values:
                (values == 0).sum()
        ),

        total_units=(
            "units",
            "sum"
        ),

        average_weekly_units=(
            "units",
            "mean"
        ),

        median_weekly_units=(
            "units",
            "median"
        ),

        max_weekly_units=(
            "units",
            "max"
        ),

        average_positive_week_units=(
            "units",
            lambda values:
                values[
                    values > 0
                ].mean()
                if (
                    values > 0
                ).any()
                else 0
        ),

        positive_week_cv=(
            "units",
            positive_week_cv
        ),

        total_revenue=(
            "revenue",
            "sum"
        ),

        total_orders=(
            "orders",
            "sum"
        ),
    )
)
```

```python
# ================================================================
# 68.7 — DEMAND OCCURRENCE METRICS
# ================================================================

product_demand_profile[
    "demand_occurrence_rate"
] = (
    product_demand_profile[
        "positive_demand_weeks"
    ]
    /
    product_demand_profile[
        "observed_weeks"
    ]
)


product_demand_profile[
    "zero_demand_rate"
] = (
    1
    -
    product_demand_profile[
        "demand_occurrence_rate"
    ]
)
```

```python
# ================================================================
# 68.8 — AVERAGE INTER-DEMAND INTERVAL
# ================================================================

def average_inter_demand_interval(
    group
):

    positive_positions = np.where(
        group[
            "units"
        ]
        .to_numpy()
        >
        0
    )[0]


    if (
        len(
            positive_positions
        )
        <
        2
    ):

        return np.nan


    return np.diff(
        positive_positions
    ).mean()


product_inter_demand = (
    product_week_panel
    .groupby(
        "product"
    )
    .apply(
        average_inter_demand_interval
    )
    .rename(
        "average_inter_demand_interval"
    )
    .reset_index()
)


product_demand_profile = (
    product_demand_profile
    .merge(
        product_inter_demand,
        on="product",
        how="left",
        validate="one_to_one"
    )
)
```

```python
# ================================================================
# 68.9 — PRODUCT HISTORY SUMMARY
# ================================================================

product_history_summary = (
    product_demand_profile[
        [
            "observed_weeks",
            "positive_demand_weeks",
            "demand_occurrence_rate",
            "average_inter_demand_interval",
            "total_units",
            "average_weekly_units",
            "positive_week_cv",
            "total_orders",
        ]
    ]
    .describe(
        percentiles=[
            0.10,
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99,
        ]
    )
)


display(
    product_history_summary
)
```

```python
# ================================================================
# 68.10 — PRODUCT DEMAND SEGMENTATION
# ================================================================

def classify_product_demand(
    row
):

    # Insufficient historical length
    if (
        row[
            "observed_weeks"
        ]
        <
        26
    ):

        return (
            "Insufficient History"
        )


    # Very limited observed demand
    if (
        row[
            "positive_demand_weeks"
        ]
        <
        8
    ):

        return (
            "Sparse / Intermittent"
        )


    occurrence = (
        row[
            "demand_occurrence_rate"
        ]
    )


    cv = (
        row[
            "positive_week_cv"
        ]
    )


    # High-frequency demand
    if (
        occurrence
        >=
        0.75
    ):

        if (
            pd.notna(
                cv
            )
            and
            cv
            <=
            1.0
        ):

            return (
                "High-Frequency Stable"
            )

        return (
            "High-Frequency Volatile"
        )


    # Moderate-frequency demand
    if (
        occurrence
        >=
        0.40
    ):

        return (
            "Medium-Frequency"
        )


    return (
        "Sparse / Intermittent"
    )


product_demand_profile[
    "demand_segment"
] = (
    product_demand_profile
    .apply(
        classify_product_demand,
        axis=1
    )
)
```

```python
# ================================================================
# 68.11 — SEGMENT SUMMARY
# ================================================================

product_segment_summary = (
    product_demand_profile
    .groupby(
        "demand_segment",
        as_index=False
    )
    .agg(
        products=(
            "product",
            "nunique"
        ),

        total_units=(
            "total_units",
            "sum"
        ),

        total_revenue=(
            "total_revenue",
            "sum"
        ),

        median_history_weeks=(
            "observed_weeks",
            "median"
        ),

        median_positive_weeks=(
            "positive_demand_weeks",
            "median"
        ),

        median_occurrence_rate=(
            "demand_occurrence_rate",
            "median"
        ),

        median_weekly_units=(
            "average_weekly_units",
            "median"
        ),
    )
)


product_segment_summary[
    "product_share"
] = (
    product_segment_summary[
        "products"
    ]
    /
    product_segment_summary[
        "products"
    ].sum()
)


product_segment_summary[
    "unit_share"
] = (
    product_segment_summary[
        "total_units"
    ]
    /
    product_segment_summary[
        "total_units"
    ].sum()
)


product_segment_summary[
    "revenue_share"
] = (
    product_segment_summary[
        "total_revenue"
    ]
    /
    product_segment_summary[
        "total_revenue"
    ].sum()
)


display(
    product_segment_summary
)
```

```python
# ================================================================
# 69.3 — PRODUCT FORECAST BASELINES
# ================================================================

eligible_product_panel[
    "forecast_last_week"
] = (
    eligible_product_panel
    .groupby(
        "product"
    )[
        "units"
    ]
    .shift(1)
)


eligible_product_panel[
    "forecast_ma4"
] = (
    eligible_product_panel
    .groupby(
        "product"
    )[
        "units"
    ]
    .transform(
        lambda values:
            values
            .shift(1)
            .rolling(
                window=4,
                min_periods=4
            )
            .mean()
    )
)


eligible_product_panel[
    "forecast_ma8"
] = (
    eligible_product_panel
    .groupby(
        "product"
    )[
        "units"
    ]
    .transform(
        lambda values:
            values
            .shift(1)
            .rolling(
                window=8,
                min_periods=8
            )
            .mean()
    )
)


eligible_product_panel[
    "forecast_seasonal_52"
] = (
    eligible_product_panel
    .groupby(
        "product"
    )[
        "units"
    ]
    .shift(52)
)
```

```python
# ================================================================
# 69.5 — PRODUCT FORECAST METRICS
# ================================================================

def evaluate_product_forecast(
    data,
    forecast_column,
    model_name
):

    actual = (
        data[
            "units"
        ]
        .to_numpy(
            dtype=float
        )
    )


    predicted = (
        data[
            forecast_column
        ]
        .to_numpy(
            dtype=float
        )
    )


    predicted = np.maximum(
        0,
        predicted
    )


    error = (
        predicted
        -
        actual
    )


    absolute_error = np.abs(
        error
    )


    mae = (
        absolute_error.mean()
    )


    rmse = np.sqrt(
        np.mean(
            error ** 2
        )
    )


    wape = (
        absolute_error.sum()
        /
        actual.sum()
        if actual.sum() > 0
        else np.nan
    )


    bias = (
        error.mean()
    )


    forecast_actual_ratio = (
        predicted.sum()
        /
        actual.sum()
        if actual.sum() > 0
        else np.nan
    )


    return pd.DataFrame(
        {
            "Model": [
                model_name
            ],

            "Product-Weeks": [
                len(
                    data
                )
            ],

            "Products": [
                data[
                    "product"
                ].nunique()
            ],

            "MAE Units": [
                mae
            ],

            "RMSE Units": [
                rmse
            ],

            "WAPE": [
                wape
            ],

            "Bias Units": [
                bias
            ],

            "Predicted / Actual Units": [
                forecast_actual_ratio
            ],
        }
    )
```

```python
# ================================================================
# 69.6 — OVERALL SKU BASELINE PERFORMANCE
# ================================================================

product_baseline_map = {
    "Last-Week Naive":
        "forecast_last_week",

    "4-Week Moving Average":
        "forecast_ma4",

    "8-Week Moving Average":
        "forecast_ma8",

    "Seasonal Naive — 52 Weeks":
        "forecast_seasonal_52",
}


product_baseline_results = []


for (
    model_name,
    forecast_column
) in (
    product_baseline_map.items()
):

    product_baseline_results.append(
        evaluate_product_forecast(
            data=
                product_forecast_evaluation,

            forecast_column=
                forecast_column,

            model_name=
                model_name
        )
    )


product_forecast_baseline_results = (
    pd.concat(
        product_baseline_results,
        ignore_index=True
    )
    .sort_values(
        "WAPE"
    )
    .reset_index(drop=True)
)


display(
    product_forecast_baseline_results
)
```

```python
# ================================================================
# 69.7 — SKU FORECAST PERFORMANCE BY SEGMENT
# ================================================================

product_segment_forecast_records = []


for segment, segment_data in (
    product_forecast_evaluation
    .groupby(
        "demand_segment"
    )
):

    for (
        model_name,
        forecast_column
    ) in (
        product_baseline_map.items()
    ):

        result = (
            evaluate_product_forecast(
                data=
                    segment_data,

                forecast_column=
                    forecast_column,

                model_name=
                    model_name
            )
        )


        result[
            "Demand Segment"
        ] = segment


        product_segment_forecast_records.append(
            result
        )


product_segment_forecast_results = (
    pd.concat(
        product_segment_forecast_records,
        ignore_index=True
    )
    .sort_values(
        [
            "Demand Segment",
            "WAPE",
        ]
    )
    .reset_index(drop=True)
)


display(
    product_segment_forecast_results
)
```

```python
# ================================================================
# 69.8 — PER-PRODUCT FORECAST PERFORMANCE
# ================================================================

sku_level_metric_records = []


for product, product_data in (
    product_forecast_evaluation
    .groupby(
        "product"
    )
):

    actual_total = (
        product_data[
            "units"
        ]
        .sum()
    )


    if (
        actual_total
        <=
        0
    ):
        continue


    segment = (
        product_data[
            "demand_segment"
        ]
        .iloc[0]
    )


    for (
        model_name,
        forecast_column
    ) in (
        product_baseline_map.items()
    ):

        absolute_error = (
            product_data[
                forecast_column
            ]
            -
            product_data[
                "units"
            ]
        ).abs().sum()


        sku_wape = (
            absolute_error
            /
            actual_total
        )


        sku_level_metric_records.append(
            {
                "product":
                    product,

                "Demand Segment":
                    segment,

                "Model":
                    model_name,

                "SKU WAPE":
                    sku_wape,

                "Actual Units":
                    actual_total,
            }
        )


sku_level_forecast_metrics = (
    pd.DataFrame(
        sku_level_metric_records
    )
)
```

```python
# ================================================================
# 69.8B — SKU-LEVEL WAPE DISTRIBUTION
# ================================================================

sku_wape_summary = (
    sku_level_forecast_metrics
    .groupby(
        "Model"
    )[
        "SKU WAPE"
    ]
    .describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
        ]
    )
)


display(
    sku_wape_summary
)
```

```python
# ================================================================
# 70.4 — CROSTON FORECAST
# ================================================================

def croston_one_step_forecasts(
    demand,
    alpha=0.10
):

    demand = np.asarray(
        demand,
        dtype=float
    )


    forecasts = np.full(
        len(demand),
        np.nan
    )


    positive_indices = np.where(
        demand > 0
    )[0]


    if len(
        positive_indices
    ) == 0:

        return forecasts


    first_positive = (
        positive_indices[0]
    )


    demand_estimate = (
        demand[
            first_positive
        ]
    )


    interval_estimate = (
        first_positive
        +
        1
    )


    last_positive_index = (
        first_positive
    )


    current_forecast = (
        demand_estimate
        /
        interval_estimate
    )


    for t in range(
        first_positive + 1,
        len(demand)
    ):

        # forecast for week t uses state available before observing week t
        forecasts[
            t
        ] = current_forecast


        if (
            demand[
                t
            ]
            >
            0
        ):

            interval = (
                t
                -
                last_positive_index
            )


            demand_estimate = (
                alpha
                *
                demand[
                    t
                ]
                +
                (
                    1
                    -
                    alpha
                )
                *
                demand_estimate
            )


            interval_estimate = (
                alpha
                *
                interval
                +
                (
                    1
                    -
                    alpha
                )
                *
                interval_estimate
            )


            last_positive_index = (
                t
            )


            current_forecast = (
                demand_estimate
                /
                interval_estimate
            )


    return forecasts
```

```python
# ================================================================
# 70.5 — SBA FORECAST
# ================================================================

def sba_one_step_forecasts(
    demand,
    alpha=0.10
):

    croston_forecasts = (
        croston_one_step_forecasts(
            demand=
                demand,

            alpha=
                alpha
        )
    )


    return (
        (
            1
            -
            alpha
            /
            2
        )
        *
        croston_forecasts
    )
```

```python
# ================================================================
# 70.6 — INTERMITTENT ALPHA CANDIDATES
# ================================================================

INTERMITTENT_ALPHA_CANDIDATES = [
    0.05,
    0.10,
    0.20,
    0.30,
]


print(
    INTERMITTENT_ALPHA_CANDIDATES
)
```

```python
# ================================================================
# 70.7 — GENERATE INTERMITTENT FORECASTS
# ================================================================

sparse_products = set(
    product_demand_profile.loc[
        product_demand_profile[
            "demand_segment"
        ]
        ==
        "Sparse / Intermittent",
        "product"
    ]
)


sparse_panel = (
    product_week_panel.loc[
        product_week_panel[
            "product"
        ]
        .isin(
            sparse_products
        )
    ]
    .copy()
    .sort_values(
        [
            "product",
            "week_start",
        ]
    )
)


intermittent_forecast_frames = []


for product, group in (
    sparse_panel
    .groupby(
        "product"
    )
):

    group = (
        group
        .copy()
        .reset_index(drop=True)
    )


    demand = (
        group[
            "units"
        ]
        .to_numpy(
            dtype=float
        )
    )


    for alpha in (
        INTERMITTENT_ALPHA_CANDIDATES
    ):

        group[
            f"croston_{alpha:.2f}"
        ] = (
            croston_one_step_forecasts(
                demand,
                alpha=
                    alpha
            )
        )


        group[
            f"sba_{alpha:.2f}"
        ] = (
            sba_one_step_forecasts(
                demand,
                alpha=
                    alpha
            )
        )


    intermittent_forecast_frames.append(
        group
    )


sparse_forecast_panel = pd.concat(
    intermittent_forecast_frames,
    ignore_index=True
)


print(
    f"Sparse products evaluated: "
    f"{sparse_forecast_panel['product'].nunique():,}"
)
```

```python
# ================================================================
# 70.8 — COMMON INTERMITTENT EVALUATION WINDOW
# ================================================================

evaluation_start = (
    product_forecast_evaluation[
        "week_start"
    ]
    .min()
)


evaluation_end = (
    product_forecast_evaluation[
        "week_start"
    ]
    .max()
)


sparse_forecast_evaluation = (
    sparse_forecast_panel.loc[
        (
            sparse_forecast_panel[
                "week_start"
            ]
            >=
            evaluation_start
        )
        &
        (
            sparse_forecast_panel[
                "week_start"
            ]
            <=
            evaluation_end
        )
    ]
    .copy()
)


# Attach MA4 / naive baselines from prior evaluation

sparse_baseline_lookup = (
    product_forecast_evaluation.loc[
        product_forecast_evaluation[
            "demand_segment"
        ]
        ==
        "Sparse / Intermittent",
        [
            "product",
            "week_period",
            "forecast_ma4",
            "forecast_last_week",
        ]
    ]
)


sparse_forecast_evaluation = (
    sparse_forecast_evaluation
    .merge(
        sparse_baseline_lookup,
        on=[
            "product",
            "week_period",
        ],
        how="inner",
        validate="one_to_one"
    )
)


print(
    f"Sparse evaluation rows: "
    f"{len(sparse_forecast_evaluation):,}"
)

print(
    f"Sparse products represented: "
    f"{sparse_forecast_evaluation['product'].nunique():,}"
)
```

```python
# ================================================================
# 70.9 — INTERMITTENT MODEL COMPARISON
# ================================================================

intermittent_results = []


# Existing baselines

for (
    model_name,
    forecast_column
) in [
    (
        "Sparse MA4",
        "forecast_ma4"
    ),
    (
        "Sparse Last-Week Naive",
        "forecast_last_week"
    ),
]:

    intermittent_results.append(
        evaluate_product_forecast(
            data=
                sparse_forecast_evaluation,

            forecast_column=
                forecast_column,

            model_name=
                model_name
        )
    )


# Croston / SBA

for alpha in (
    INTERMITTENT_ALPHA_CANDIDATES
):

    for method in [
        "croston",
        "sba",
    ]:

        column = (
            f"{method}_{alpha:.2f}"
        )


        valid_data = (
            sparse_forecast_evaluation
            .dropna(
                subset=[
                    column
                ]
            )
        )


        result = (
            evaluate_product_forecast(
                data=
                    valid_data,

                forecast_column=
                    column,

                model_name=
                    (
                        f"{method.upper()} "
                        f"alpha={alpha:.2f}"
                    )
            )
        )


        intermittent_results.append(
            result
        )


intermittent_forecast_results = (
    pd.concat(
        intermittent_results,
        ignore_index=True
    )
    .sort_values(
        "WAPE"
    )
    .reset_index(drop=True)
)


display(
    intermittent_forecast_results
)
```

```python
# ================================================================
# 70.10 — INTERMITTENT SKU-LEVEL WAPE
# ================================================================

intermittent_candidate_columns = {
    "Sparse MA4":
        "forecast_ma4",

    "Sparse Last-Week Naive":
        "forecast_last_week",
}


for alpha in (
    INTERMITTENT_ALPHA_CANDIDATES
):

    intermittent_candidate_columns[
        f"CROSTON alpha={alpha:.2f}"
    ] = (
        f"croston_{alpha:.2f}"
    )


    intermittent_candidate_columns[
        f"SBA alpha={alpha:.2f}"
    ] = (
        f"sba_{alpha:.2f}"
    )


intermittent_sku_records = []


for product, product_data in (
    sparse_forecast_evaluation
    .groupby(
        "product"
    )
):

    actual_total = (
        product_data[
            "units"
        ]
        .sum()
    )


    if (
        actual_total
        <=
        0
    ):
        continue


    for (
        model_name,
        forecast_column
    ) in (
        intermittent_candidate_columns.items()
    ):

        valid = (
            product_data[
                forecast_column
            ]
            .notna()
        )


        if (
            valid.sum()
            ==
            0
        ):
            continue


        actual = (
            product_data.loc[
                valid,
                "units"
            ]
        )


        predicted = (
            product_data.loc[
                valid,
                forecast_column
            ]
        )


        denominator = (
            actual.sum()
        )


        if (
            denominator
            <=
            0
        ):
            continue


        sku_wape = (
            (
                predicted
                -
                actual
            )
            .abs()
            .sum()
            /
            denominator
        )


        intermittent_sku_records.append(
            {
                "product":
                    product,

                "Model":
                    model_name,

                "SKU WAPE":
                    sku_wape,
            }
        )


intermittent_sku_metrics = pd.DataFrame(
    intermittent_sku_records
)


intermittent_sku_wape_summary = (
    intermittent_sku_metrics
    .groupby(
        "Model"
    )[
        "SKU WAPE"
    ]
    .describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
        ]
    )
)


display(
    intermittent_sku_wape_summary
)
```

## Final Forecasting Limitations

The final forecasting system is subject to several important limitations.

### Limited Historical Depth

The dataset contains approximately two years of weekly history.

Annual seasonality is therefore supported by only a limited number of repeated seasonal cycles.

### No Pristine Final Weekly Holdout

The eligible later periods were inspected during baseline diagnostics.

Final model evidence therefore relies primarily on rolling-origin and multi-horizon temporal backtesting rather than an untouched final Test period.

### Aggregate Forecasting Is Stronger Than SKU Forecasting

Business-level annual seasonality is substantially more stable than product-level seasonality.

The products generating seasonal demand change materially over time.

### SKU Forecasts Are Planning Signals

Individual product forecasts contain substantial error and should not be interpreted as precise purchase-order quantities without additional inventory controls.

### Inventory Availability Is Unknown

The dataset records realised sales rather than unconstrained latent demand.

Observed zero or low sales may sometimes reflect product availability rather than true customer demand.

No stock-on-hand, stock-out or replenishment records are available to distinguish those cases.

### Promotions and Price Changes Are Not Explicitly Modelled

Future promotional calendars, merchandising actions and planned price changes are not available as forward-looking model inputs.

### External Conditions Are Not Modelled

Macroeconomic changes, competitor activity, supply disruption and other external events may cause demand to diverge from historical patterns.

### Forecast Ranges Are Empirical

Historical planning ranges are derived from observed backtest residuals.

They are not independently calibrated probabilistic guarantees.
